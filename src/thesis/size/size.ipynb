```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("finite_size_data.csv")

plt.figure(figsize=(10, 6))

markers = ['o', 's', '^']
for i, n_val in enumerate(sorted(df['N'].unique())):
    subset = df[df['N'] == n_val].sort_values('T')
    plt.plot(subset['T'], subset['success_rate'], 
             marker=markers[i], label=f'N={n_val}')

plt.xscale('log')
plt.xlabel('Annealing Steps $T$')
plt.ylabel('Success Probability')
plt.title('Success Probability vs. Annealing Speed (Finite Size Effect)')
plt.legend()
plt.grid(True, which="both", ls="-", alpha=0.3)
plt.show()
```
