```python
#non annealing

import pandas as pd
import matplotlib.pyplot as plt

# Parameter
seed = 42
iter = 1e+04
tau = 100
beta = 1e-2
file_name = f'nonannealing_tau{tau:.0e}_beta{beta:.0e}_energy'

file_path_energy = f'/home/mori-lab/shimizu/aco/aco-on-spinglass/src/sampling/python_alpha_annealing/data/seed{seed}/iter{iter:.0e}/{file_name}.csv'

data = pd.read_csv(file_path_energy)
data = data['energy']
min = data.min()
print(min)

plt.plot(data, '.')
#plt.plot(data_alpha)
#plt.xscale('log')
plt.xlabel('iter')
plt.ylabel('energy')
plt.title(f'α-annealing, min={min:.3f}')
#plt.xlim(0, 1e2)
plt.show()
```
