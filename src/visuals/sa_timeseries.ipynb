```python
import pandas as pd
import matplotlib.pyplot as plt

# Parameter
seed = 42
iter = 1e+07
initial_temperature = 100
goal_temperature = 0.5

file_path = f'/home/mori-lab/shimizu/aco/aco-on-spinglass/src/sampling/simulated_annealing/data/seed{seed}/iter{iter:.0e}/it{initial_temperature}_gt{goal_temperature}.csv'

data = pd.read_csv(file_path)

plt.plot(data)
plt.xscale('log')
plt.xlabel('iteration')
plt.ylabel('energy')
plt.xlim(1e+3, iter)
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Parameter
seed = 42
iter = 1e+07
initial_temperature = 100
goal_temperature = 0.01

file_path = f'/home/mori-lab/shimizu/aco/aco-on-spinglass/src/sampling/simulated_annealing/data/seed{seed}/iter{iter:.0e}/it{initial_temperature}_gt{goal_temperature}.csv'

data = pd.read_csv(file_path)
data = data['energy']
min = data.min()

plt.plot(data)
plt.title(f'SA, min={min:.3f}')
plt.xscale('log')
plt.xlabel('iter')
plt.ylabel('energy')
plt.xlim(1e+3, iter)
plt.show()
print(min)
```
