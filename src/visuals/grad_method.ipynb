```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
# Parameter
seed = 42
iter = 1e+07
alpha = 1e-03
lr = 1e-03
tol = 1e-06
file_name = f'alpha{alpha:.0e}_lr{lr:.0e}_tol{tol:.0e}'

file_path = f'/home/mori-lab/shimizu/aco/aco-on-spinglass/src/sampling/grad_method/data/seed{seed}/iter{iter:.0e}/{file_name}.csv'

data = pd.read_csv(file_path)
data_energy = data['energy']
min = data_energy.min()

plt.plot(data_energy)
plt.xscale('log')
plt.xlim(1e5, 1e6)
plt.xlabel('iteration')
plt.ylabel('energy')
plt.title(f'lr={lr:.0e}, min={min:.3f}')
plt.show()
print(int(np.sum(data['alpha'])))
print(min)
```
