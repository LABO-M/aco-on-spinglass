```python
from pathlib import Path
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline

ROOT = Path.home() / "shimizu/aco-on-spinglass/src/sampling/return_M/compare_decision"
DATA = ROOT / "data"

def load_csv(p):
    x = np.loadtxt(p)
    return x[np.isfinite(x)]

def show_hist(path, alpha, bins=120):
    M = load_csv(path)
    M = M[(M>=-alpha) & (M<=alpha)]
    plt.figure(figsize=(8,5))
    plt.hist(M, bins=bins, range=(-alpha,alpha), density=True)
    plt.xlim(-alpha, alpha)
    plt.xlabel("m"); plt.ylabel("p(m)")
    plt.title(path.name)
    plt.tight_layout()
    plt.show()

# 例：alpha=0.835... の比較
alpha = 0.835118240479
for p in sorted(DATA.glob(f"*alpha{alpha}*.csv")):
    show_hist(p, alpha)
```
